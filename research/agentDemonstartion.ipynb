```python
import os
import certifi
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_core.tools import tool
import requests
```

```python
from langchain.agents import create_react_agent,AgentExecutor
from langchain import hub
```

```python
#Load Env variable
os.environ["SSL_CERT_FILE"]=certifi.where()
load_dotenv()
openapi_api_key=os.getenv("open_api_key")
groq_api_key=os.getenv("groq_api_key")
tavily_api_key=os.getenv("tavily_api_key")
```

```python
search_tool=TavilySearchResults(max_results=2)
```

```python
result=search_tool.invoke("What is the capitl of frnace and it's total population")
result
```

```python
#LLM
# llm=ChatOpenAI(
#     model="gpt-4o-mini",
#     temperature=0,
#     api_key=openapi_api_key
# )
groq_llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0,
    api_key=groq_api_key
)
```

```python

weather_api_key = os.getenv("weather_api_key")


@tool
def get_weather(city: str) -> str:
    """Get the current weather for a city.
    Use this tool when the user asks about weather,
    temperature, humidity, or weather conditions.
    The input must be a city name.
    """

    url = "https://api.weatherstack.com/current"

    params = {
        "access_key": weather_api_key,
        "query": city,
        "units": "m"
    }

    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    data = response.json()

    # Weatherstack can return an error object
    if "error" in data:
        return f"Unable to get weather: {data['error']['info']}"

    location = data["location"]["name"]
    temperature = data["current"]["temperature"]
    humidity = data["current"]["humidity"]
    description = data["current"]["weather_descriptions"][0]

    return (
        f"Weather in {location}: "
        f"{description}, "
        f"Temperature: {temperature}°C, "
        f"Humidity: {humidity}%"
    )
```

```python
prompt=hub.pull("hwchase17/react")
```

```python
tools=[search_tool,get_weather]
```

```python
agent=create_react_agent(
    llm=groq_llm,
    tools=tools,
    prompt=prompt
)
```

```python
agent_executor=AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True
)
```

```python
response1 = agent_executor.invoke(
    {
        "input":(
            "Tell me about Iran and America latest war"
            "Tell me the Today temperature of new delhi"
            
        )
    }
)
```
